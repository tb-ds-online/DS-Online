## Cómo funciona el juego

Es una versión simplificada del juego original, pensada para que el desarrollo sea abordable:

1. Hay dos jugadores: **vosotros y la máquina**.
2. Cada jugador tiene un **tablero de 10 × 10** casillas.
3. Al empezar se colocan los barcos de los dos tableros **de forma aleatoria**. Los barcos son:
    * 4 barcos de 1 casilla de eslora
    * 3 barcos de 2 casillas de eslora
    * 2 barcos de 3 casillas de eslora
    * 1 barco de 4 casillas de eslora

    En total, **20 casillas con barco** por tablero.
4. Se juega **por turnos y empezáis vosotros**.
5. En cada turno se dispara a una coordenada **(fila, columna)** del tablero rival. Las filas y las columnas van **del 0 al 9**.
6. **Si el disparo acierta, se vuelve a disparar**. Si da en agua, pasa el turno al rival. Esto vale también para la máquina.
7. La máquina dispara a una **casilla aleatoria** de vuestro tablero.
8. Cuando un jugador tiene **todos sus barcos hundidos** (sus 20 casillas tocadas), la partida termina y gana el otro.

Además, para que el juego no se pueda romper:

* **Disparo repetido**: si disparáis a una casilla a la que ya habíais disparado, no cuenta. El programa avisa y vuelve a pedir coordenadas. La máquina **nunca** repite casilla.
* **Entrada no válida**: si se escribe algo que no es una coordenada (letras, números fuera del 0–9, un solo número…), el programa avisa y vuelve a pedirla. El programa no se cierra con un error.

Para ver cómo es el juego original, en [esta página](http://es.battleship-game.org/) podéis jugar online.

## Qué tenéis que construir

El repositorio tendrá, como mínimo, esta estructura:

```
hundir-la-flota/
├── main.py            # arranca y ejecuta la partida
├── clases.py          # la clase Tablero
├── funciones.py       # funciones auxiliares del juego
├── variables.py       # constantes
├── requirements.txt   # dependencias (numpy)
├── README.md          # cómo se ejecuta y cómo se juega
├── AI_LOG.md          # registro de uso de IA
└── .gitignore
```

### 1. `variables.py`: las constantes

Todo lo que no cambia durante la partida: el tamaño del tablero, los barcos y su eslora, los símbolos que vais a pintar y los mensajes fijos. Por ejemplo, `TAMANO_TABLERO = 10`.

Ningún otro archivo debería tener un `10` escrito a mano: si mañana el tablero pasa a ser de 12 × 12, solo se cambia aquí.

### 2. `clases.py`: la clase `Tablero`

Cada jugador tiene su propio objeto `Tablero`. Al crearlo, el constructor recibe o define:

* Un **id de jugador**, para saber de quién es el tablero.
* Las **dimensiones** del tablero, que salen de vuestras constantes.
* Los **barcos** que hay que colocar. Basta con un diccionario con el nombre de cada barco y su eslora.
* **Dos arrays de NumPy** de 10 × 10:
    * `tablero`: los barcos de ese jugador y los disparos que ha recibido. Es lo que veis cuando imprimís **vuestro** tablero.
    * `disparos`: solo los disparos recibidos, **sin barcos**. Es lo que ve el rival. Así podéis enseñar el tablero de la máquina sin descubrir dónde tiene los barcos.

Y como mínimo estos métodos (los nombres son los que usa la guía de GitHub para los issues):

| Método | Qué hace |
|---|---|
| `colocar_barcos()` | Coloca todos los barcos en el tablero. Cada barco tiene que quedar **entero dentro del tablero** y **sin pisar a otro barco**. No hace falta dejar agua entre barcos. |
| `recibir_disparo(fila, columna)` | Comprueba si en esa casilla hay barco o agua, lo **marca en los dos arrays** e indica el resultado: tocado, agua o repetido. |
| `mostrar()` | Imprime el tablero. Con una opción para ocultar los barcos, que es la que usaréis con el tablero de la máquina. |

No os limitéis a estos métodos: cread todos los que os hagan falta (por ejemplo, uno que diga si a un jugador le quedan barcos).

### 3. `funciones.py`: el resto de la lógica

Todas las funciones que no son métodos de `Tablero`: pedir y validar las coordenadas del jugador, elegir el disparo de la máquina sin repetir casilla, imprimir las instrucciones…

### 4. `main.py`: la partida

El programa que se ejecuta con `python main.py`:

1. Muestra un mensaje de bienvenida y las instrucciones.
2. Crea los dos tableros y coloca los barcos. Esto pasa **una sola vez**, al principio.
3. Entra en un bucle `while True:` que repite los turnos:
    * Al empezar vuestro turno, se imprime **vuestro tablero** (para ver qué os ha tocado la máquina) y el **tablero de la máquina sin barcos** (para ver dónde habéis disparado ya).
    * Se piden coordenadas y se dispara al tablero de la máquina.
        * **Tocado**: se marca y volvéis a disparar.
        * **Agua**: se marca y dispara la máquina, que sigue disparando mientras acierte.
    * Cuando a un jugador no le quedan barcos, se anuncia quién ha ganado y se sale del bucle con `break`.

Hay muchas formas de resolver este proyecto. Mientras cumpla las reglas y la estructura, elegid la que os resulte más cómoda.

## Recomendaciones

1. **Empezad con los barcos en posiciones fijas.** Colocarlos al azar es la parte más difícil. Con un tablero fijo podéis programar y probar todo el juego, y meter la colocación aleatoria cuando lo demás funcione. No es como funciona el juego, pero avanzar es mejor que atascarse el primer día.
2. **Colocación aleatoria**: elegid una casilla de inicio al azar y una orientación **N, S, E u O**. Un barco de eslora 3 con orientación N ocupa la casilla de inicio y las dos de encima. Antes de colocarlo, comprobad que todas sus casillas están dentro del tablero y libres; si no, probad con otra casilla.
3. **No hace falta una clase `Barco`**. Basta con saber cuántas casillas con barco quedan sin tocar: con 20 casillas con barco, es como tener 20 vidas.
4. **Elegid bien los símbolos.** Al imprimir el tablero se tienen que distinguir cuatro casos:
    * Agua sin disparar
    * Agua donde se ha disparado
    * Barco sin tocar
    * Barco tocado
5. **Usad el debugger de VS Code** si el programa no hace lo que esperáis. Un breakpoint y el panel Variables os dicen qué hay en cada array en ese momento.
6. **`continue` y `break`** os ayudan a controlar el bucle: `continue` para volver a pedir coordenadas tras una entrada no válida, `break` para terminar la partida.

### Si os sobra tiempo

Ideas para ir más allá, solo cuando el juego base funcione:

* Avisar de **«tocado y hundido»** cuando se termina de hundir un barco.
* Escribir `salir` para terminar la partida en cualquier momento.
* Colocar los barcos **sin que se toquen**, ni siquiera en diagonal.
* Una máquina más lista: cuando acierta, que dispare a las casillas de alrededor.
